```python
import pandas as pd
import numpy as np
pd.options.display.float_format = '{:.2f}'.format

#tmass_psc_xsc_best_neighbour

T22 = pd.read_csv('../data/Tailo_2022_table_1.csv', dtype={'GaiaeDR3 ID': 'str'})[0:32]

stars = list(T22['GaiaeDR3 ID'].astype(int).astype(str))
cols = ['source_id', 'logg_gspphot', 'logg_gspphot_lower', 'logg_gspphot_upper', 'teff_gspphot', 'teff_gspphot_lower', 'teff_gspphot_upper', 'mh_gspphot', 'mh_gspphot_lower', 'mh_gspphot_upper',\
        'distance_gspphot', 'distance_gspphot_lower', 'distance_gspphot_upper', 'ag_gspphot', 'ag_gspphot_lower', 'ag_gspphot_upper',\
        'ebpminrp_gspphot', 'ebpminrp_gspphot_lower', 'ebpminrp_gspphot_upper', 'mg_gspphot','mg_gspphot_lower', 'mg_gspphot_upper',\
        'teff_gspspec','teff_gspspec_lower', 'teff_gspspec_upper', 'logg_gspspec','logg_gspspec_lower', 'logg_gspspec_upper', 'mh_gspspec',\
        'mh_gspspec_lower', 'mh_gspspec_upper', 'alphafe_gspspec','alphafe_gspspec_upper','alphafe_gspspec_lower']

cols = [f'cat.{i}' for i in cols]
gaiacat = "gaiadr3.astrophysical_parameters"  # or whichever you prefer

# If you are using the gaia module and not TapPlus, ignore this


def adql_escape(star):
    """
    A function that can make stings in the right format for adql queries.
    Inspiration from https://github.com/
    gmantele/taplib/blob/master/src/adql/query/operand/StringConstant.java
    """
    return "'" + star.replace("'", "''") + "'"

print(f"""    
SELECT {", ".join(cols)}
FROM {gaiacat} AS cat
WHERE cat.source_id
IN ({",".join(adql_escape(s) for s in stars)})
                """,)
```
