```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob
```

```python
files = glob.glob('../data/M4RGB_ps/M4*_PS.txt')


for i in range(len(files)):

    ID = files[i].split('/')[-1].split('_PS.txt')[0]
    
    fig = plt.figure(figsize = (8, 3))

    spect = pd.read_csv(files[i], names = ['freq', 'power'], sep = '\s+')

    plt.loglog(spect['freq'], spect['power'])

    plt.xlabel(r'Frequency [$\mu$Hz]')

    plt.ylabel(r'Power [ppm$^2\mu$Hz$^{-1}$]')

    fig.savefig(f'../figs/spectra/{ID}.png')

    fig.clf()
```
