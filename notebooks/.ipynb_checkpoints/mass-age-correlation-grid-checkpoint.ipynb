```python
import pandas as pd
import numpy as np
import os
from pathlib import Path
import pickle
import time
from itertools import product

#### graphical
import matplotlib.pyplot as plt
import corner

```

```python
df_full = pd.read_hdf('../grids/Chiara.hdf5', key='df')

df_test = df_full.sample(10000)
```

```python
def massfromage(age, A = 0.677):
    return (A*10**4/age)**(1/3.2)
```

```python
test_age = np.linspace(np.min(df_test['age']) - 3000, np.max(df_test['age']), 1000)

plt.scatter(df_test['age']/10**3, df_test['massini'], facecolor = 'skyblue', alpha = 0.3)
plt.plot(test_age/10**3, massfromage(test_age, A = np.median(df_test['age']*df_test['massini']**3.2/10**4)), color = 'deeppink', linestyle = '--', linewidth = 3)
plt.plot(test_age/10**3, massfromage(test_age + 3000, A = np.median(df_test['age']*df_test['massini']**3.2/10**4)), color = 'deeppink', linewidth = 3, alpha = 0.5)
plt.plot(test_age/10**3, massfromage(test_age - 3000, A = np.median(df_test['age']*df_test['massini']**3.2/10**4)), color = 'deeppink', linewidth = 3, alpha = 0.5)
plt.ylim(0.7, 1.5)
plt.xlim(0, 20)
plt.xlabel('Age [Gyr]')
plt.ylabel(r'Mass [M$_{\odot}$]')
plt.savefig('./mass-age-scaling-with-spread-A-0.64.png', bbox_inches = 'tight', dpi = 150)
```
