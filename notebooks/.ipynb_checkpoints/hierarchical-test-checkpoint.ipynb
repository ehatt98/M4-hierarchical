```python
def numpyro_model(uncs, obs = None):
#    diag = jnp.array(uncs)
    #define distributions to sample model parameters from, these then need to be normalised

    mu_zini = numpyro.sample('mu_zini', dist.TruncatedNormal(0.014, 0.01, low = 0, high = 0.06))

    std_zini = numpyro.sample('std_zini', dist.TruncatedNormal(0.01, 0.01, low =1e-7, high = 0.01))

    mu_alphaMLT = numpyro.sample('mu_alphaMLT', dist.TruncatedNormal(1.95, 0.1, low = 1.5, high = 2.3))

    std_alphaMLT = numpyro.sample('std_alphaMLT', dist.TruncatedNormal(0.1, 0.1, low = 1e-7, high = 1))

    mu_age = numpyro.sample('mu_age', dist.TruncatedNormal(10000, 3000, low = 1000, high = 20000))

    std_age = numpyro.sample('std_age', dist.TruncatedNormal(3000, 3000, low = 10, high = 10000))

    with numpyro.plate("star", len(uncs)):

            massini = numpyro.sample('massini', dist.TruncatedNormal(1, 0.1, low = 0.7, high = 1.5))
    
            zini = numpyro.sample('zini', dist.TruncatedNormal(mu_zini, std_zini, low = 0, high = 0.06))

            yini = numpyro.sample('yini', dist.TruncatedNormal(0.28, 0.1, low = 0.22, high = 0.35))
    
            alphaMLT = numpyro.sample('alphaMLT', dist.TruncatedNormal(mu_alphaMLT, std_alphaMLT, low = 1.5, high = 2.3))

            age = numpyro.sample('age', dist.TruncatedNormal(mu_age,std_age, low = 1000, high = 20000))

            eta = numpyro.sample('eta', dist.TruncatedNormal(0.15, 0.1, low = 0, high = 0.3))

            alphaFe = numpyro.sample('alphaFe', dist.TruncatedNormal(0.2, 0.3, low = -0.2, high =0.6))

            params_norm = jnp.stack([massini, zini, yini, alphaMLT, age, eta, alphaFe], axis = -1)

            pred_norm = model_outputs(params_norm)

#            LPhot_out = numpyro.deterministic("LPhot_out", 10**pred_norm[:, 1])

    y = numpyro.sample("y", dist.Normal(pred_norm, uncs), obs=obs)
    

#    y = numpyro.sample("y", dist.Normal(pred_norm, uncs), obs=obs)
    #unpack 
#    if obs is None:

#        FeH = numpyro.sample("FeH", dist.Normal(pred_norm[:, 0], uncs[:, 0]), obs=None)

#        LPhot = numpyro.sample("LPhot", dist.Normal(LPhot_out, uncs[:, 1]), obs=None)

#        Teff = numpyro.sample("Teff", dist.Normal(pred_norm[:, 2], uncs[:, 2]), obs=None)
        
#        numax = numpyro.sample("numax", dist.Normal(pred_norm[:, 3], uncs[:, 3]), obs=None)

#        dnuSer = numpyro.sample("dnuSer", dist.Normal(pred_norm[:, 4], uncs[:, 4]), obs=None)

#    else:
    
#        FeH = numpyro.sample("FeH", dist.Normal(pred_norm[:, 0], uncs[:, 0]), obs=jnp.array(obs[:, 0]))
    
#        lPhot = numpyro.sample("LPhot", dist.Normal(LPhot_out, uncs[:, 1]), obs=jnp.array(obs[:, 1]))

#        Teff = numpyro.sample("Teff", dist.Normal(pred_norm[:, 2], uncs[:, 2]), obs=jnp.array(obs[:, 2]))

#        numax = numpyro.sample("numax", dist.Normal(pred_norm[:, 3], uncs[:, 3]), obs=jnp.array(obs[:, 3]))

#        dnuSer = numpyro.sample("dnuSer", dist.Normal(pred_norm[:, 4], uncs[:, 4]), obs=jnp.array(obs[:, 4]))
    
```

# Prior Predictive Test

```python
phys_unc = {'Teff':150, 'LPhot':4, 'numax': 10/3090, 'dnuSer':1/135, 'FeH':0.1}
uncert = jnp.array([phys_unc[i] for i in outputs])*jnp.ones_like(jnp.array(df_obs[outputs]))

df_obs[outputs], uncert
```

```python
rng_key = jax.random.PRNGKey(0)

rng_key, rng_key_ = jax.random.split(rng_key)

prior_predictive = Predictive(numpyro_model, num_samples=1000,)

prior = prior_predictive(rng_key_, \
                         uncs=uncert,)

#y = prior.pop("y")
#for i, key in enumerate(model_outputs.outputs):
#    prior[key] = y[..., i]
coords = {"star": np.arange(num_stars)}
dims = {}
for key, value in prior.items():
    dims[key] = dim = []
    if value.ndim == 1:
        continue
    if value.shape[-1] == num_stars:
        dim.append("star")

data = az.from_numpyro(prior=prior, coords=coords, dims=dims)
data
```

# Now to Fit

```python
num_warmup, num_samples = 1000, 4000

num_chains = 2

sampler = NUTS(numpyro_model, target_accept_prob=0.8, find_heuristic_step_size=True, init_strategy=numpyro.infer.init_to_median) #May be worth trying init_to_sample

mcmc = MCMC(sampler, num_warmup=num_warmup, num_samples=num_samples, num_chains=num_chains)
```

```python
mcmc.run(jax.random.PRNGKey(0),  uncs=uncert, obs=jnp.array(df_obs[outputs]))

mcmc.print_summary()
```

```python
df_obs
```

Now we do a posterior predictive check to make sure everything makes sense

Plotting with arviz

```python
posterior = mcmc.get_samples(group_by_chain=True)
sample_stats = mcmc.get_extra_fields(group_by_chain=True)

rng, key = jax.random.split(jax.random.PRNGKey(0))
posterior_predictive = Predictive(numpyro_model, posterior_samples=posterior, parallel=True, return_sites=["y"], batch_ndims=2)
y = posterior_predictive(key,  obs=None, \
                         uncs=uncert, )["y"]

for i, key in enumerate(model_outputs.outputs):
    posterior[key] = y[..., i]

sample_stats = mcmc.get_extra_fields(group_by_chain=True)

data += az.from_dict(posterior, sample_stats=sample_stats, coords=coords, dims=dims)
data
```

```python
stat_funcs = {
    "q_16": lambda x: np.quantile(x, .16),
    "q_50": lambda x: np.quantile(x, .5),
    "q_84": lambda x: np.quantile(x, .84),
}

az.summary(data, fmt="xarray", stat_funcs=stat_funcs)
az.plot_trace(data, var_names=['mu_zini', 'mu_alphaMLT', 'mu_age'])
```
